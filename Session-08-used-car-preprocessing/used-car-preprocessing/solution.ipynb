```python
df["Mileage"] = pd.to_numeric(
    df["Mileage"].str.replace(" kmpl", "", regex=False),
    errors="coerce"
)
```

```python
df["Mileage"]
```

```python
df["Engine"] = pd.to_numeric(
    df["Engine"].str.replace(" CC", "", regex=False),
    errors="coerce"
)
```

```python
df["Engine"]
```

```python
df["Power"] = pd.to_numeric(
    df["Power"].str.replace(" bhp", "", regex=False),
    errors="coerce"
)
```

```python
df["Power"]
```

```python
df["Mileage"] = df["Mileage"].fillna(df["Mileage"].median())
```

```python
df["Engine"] = df["Engine"].fillna(df["Engine"].median())
```

```python
df["Power"] = df["Power"].fillna(df["Power"].median())
```

```python
df["Seats"] = df["Seats"].fillna(df["Seats"].median())
```

```python
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
```

```python
df["Fuel_Type"] = df["Fuel_Type"].fillna(df["Fuel_Type"].mode()[0])
```

```python
df["Transmission"] = df["Transmission"].fillna(df["Transmission"].mode()[0])
```

```python
df["Owner_Type"] = df["Owner_Type"].fillna(df["Owner_Type"].mode()[0])
```

```python
df = df.drop(columns=["New_Price"])
```

```python
df = df.dropna(subset=["Price"])
```

```python
df = df.drop(columns=["S.No.", "Name"])
```

```python
df["Owner_Type"] = df["Owner_Type"].replace({
    "First": 1,
    "Second": 2,
    "Third": 3,
    "Fourth": 4,
    "Fourth & Above": 4
})
```

```python
df = pd.get_dummies(
    df,
    columns=["Location"],
    drop_first=True,
    dtype=int
)
```

```python
df = pd.get_dummies(
    df,
    columns=["Fuel_Type"],
    drop_first=True,
    dtype=int
)
```

```python
df = pd.get_dummies(
    df,
    columns=["Transmission"],
    drop_first=True,
    dtype=int
)
```

# **The dataset was first loaded and checked for missing values and data types. Numerical columns such as Mileage, Engine, Power, and Seats were cleaned and missing values were replaced using the median because car data can contain extreme values. Categorical columns were filled using the mode because it is a simple way to replace missing categories. New_Price was removed because it contains a large percentage of missing values, while rows with missing Price were removed because Price is the target variable for the pricing model. Owner_Type was encoded using ordered numbers because the ownership categories have a natural order. Location, Fuel_Type, and Transmission were one-hot encoded because they are categorical variables without a natural numerical order.**
