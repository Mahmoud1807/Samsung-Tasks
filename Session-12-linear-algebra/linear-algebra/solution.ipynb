```python
import numpy as np
import pandas as pd


# =========================
# Task 1
# =========================

A = np.array([4.2, 1, 0.87, 3.9])
B = np.array([3.8, 1, 0.91, 4.1])

M1 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

M2 = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])


# 1.a) Dot product
# I used both methods just to check that they give the same result.

dot1 = np.dot(A, B)
dot2 = A @ B

print("Task 1.a")
print("Using np.dot():", dot1)
print("Using @:", dot2)
print("Same result:", np.isclose(dot1, dot2))


# 1.b) Magnitude of A and B

print("\nTask 1.b")

magnitude_A = np.linalg.norm(A)
magnitude_B = np.linalg.norm(B)

print("Magnitude of A:", magnitude_A)
print("Magnitude of B:", magnitude_B)


# 1.c) Matrix multiplication

print("\nTask 1.c")

result1 = M1 @ M2

print("M1 @ M2:")
print(result1)
print("Shape:", result1.shape)



result2 = M2 @ M1

print("\nM2 @ M1:")
print(result2)
print("Shape:", result2.shape)



# 1.d) Transpose

print("\nTask 1.d")

M1_transpose = M1.T

print("M1 transpose:")
print(M1_transpose)

square_result = M1 @ M1_transpose

print("\nM1 @ M1.T:")
print(square_result)

print("Shape:", square_result.shape)

```

```python

# =========================
# Task 2
# =========================

print("\n\nTask 2")


# 2.a) One-hot encoding

categories = [
    "Electronics",
    "Fashion",
    "Home",
    "Electronics"
]

encoded_categories = pd.get_dummies(categories)

print("\nTask 2.a")
print(encoded_categories)


# 2.b) Ratings matrix

R = np.array([
    [5, 0, 3, 1],
    [4, 0, 0, 1],
    [1, 1, 0, 5]
])

# I chose these weights for the two latent dimensions.
# First column = value-seeking
# Second column = trend-following

W = np.array([
    [0.9, 0.2],
    [0.3, 0.8],
    [0.7, 0.4],
    [0.5, 0.9]
])

scores = R @ W

print("\nTask 2.b")
print("R @ W:")
print(scores)


# 2.c) Find the user with the highest score

def top_user_by_dimension(scores, dim_index):
    return np.argmax(scores[:, dim_index])


best_value_user = top_user_by_dimension(scores, 0)
best_trend_user = top_user_by_dimension(scores, 1)

print("\nTask 2.c")

print("User with highest value-seeking score:",
      best_value_user)

print("User with highest trend-following score:",
      best_trend_user)

```

```python

# =========================
# Task 3
# =========================

print("\n\nTask 3")


X = np.array([
    [1, 0.10, 2],
    [1, 0.25, 5],
    [1, 0.05, 1],
    [1, 0.40, 8]
])

Y = np.array([
    [120],
    [340],
    [90],
    [510]
])


# 3.a) Calculate X transpose multiplied by X

XtX = X.T @ X

print("\nTask 3.a")
print("X.T @ X:")
print(XtX)


# 3.b) Check if the matrix can be inverted

det = np.linalg.det(XtX)
condition = np.linalg.cond(XtX)

print("\nTask 3.b")
print("Determinant:", det)
print("Condition number:", condition)

# The determinant is not zero, so the matrix is invertible.
# Also, the condition number is not extremely large,
# so I consider it safe enough to invert for this example.


# 3.c) Calculate beta in two ways

beta1 = np.linalg.inv(XtX) @ X.T @ Y

beta2 = np.linalg.solve(XtX, X.T @ Y)

print("\nTask 3.c")

print("Beta using inverse:")
print(beta1)

print("\nBeta using solve:")
print(beta2)

print("\nAre they approximately equal?")
print(np.allclose(beta1, beta2))
```
