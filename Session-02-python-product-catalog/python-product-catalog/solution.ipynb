#**Task 2**

```python
products = [
    {"name": "Galaxy Buds", "price": 89.99, "qty": 12, "category": "Audio"},
    {"name": "SmartWatch S", "price": 249.50, "qty": 3, "category": "Wearable"},
    {"name": "Power Bank 20K", "price": 34.90, "qty": 40, "category": "Accessory"}
]

total_value = sum([p["price"] * p["qty"] for p in products])
print("\nTotal Inventory Value = ${:.2f}".format(total_value))
```

```python
print("\nNumbered Catalog")
for index, product in enumerate(products, start=1):
    print(f"{index}. {product['name']} - ${product['price']:.2f}")
```

```python
print("Products Above $50")
expensive_products = [p for p in products if p["price"] > 50]

for product in expensive_products:
    print(f"{product['name']} - ${product['price']:.2f}")
```

```python
print("Products Sorted by Price (Highest First)")
sorted_products = sorted(products, key=lambda p: p["price"], reverse=True)

for product in sorted_products:
    print(f"{product['name']} - ${product['price']:.2f}")
```

#**Task3**

```python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def apply_discount(self, percent):
        self.price -= self.price * (percent / 100)


class Electronics(Product):
    def __init__(self, name, price, quantity, warranty_months):
        super().__init__(name, price, quantity)
        self.warranty_months = warranty_months

    def __str__(self):
        return (f"{self.name} | Price: ${self.price:.2f} | "
                f"Quantity: {self.quantity} | "
                f"Warranty: {self.warranty_months} months")


galaxy_buds = Electronics("Galaxy Buds", 89.99, 12, 12)
smartwatch = Electronics("SmartWatch S", 249.50, 3, 24)

galaxy_buds.apply_discount(10)
smartwatch.apply_discount(5)

print("Electronics Products")
print(galaxy_buds)
print(smartwatch)
```
